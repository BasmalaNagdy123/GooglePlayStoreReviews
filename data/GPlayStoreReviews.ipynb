```python
df_analysis=df.sample(500,random_state=40)
```

```python
df_analysis
```

```python
import re
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('all')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
def clean_cont(content):
  cont=content.lower()
  cont=re.sub(r'[^a-z\s]',' ',cont)
  tokens=word_tokenize(cont)
  tokens=[i for i in tokens if i not in stop_words]
  return ' '.join(tokens)
```

```python
df_analysis['cleand_content']=df_analysis['content'].apply(clean_cont)
```

```python
df_analysis.head(2)
```

```python
from textblob import TextBlob
```

```python
def sentiment(content):
  p=TextBlob(content).sentiment.polarity
  if p>0:
    return 'Positive'
  elif  p<0:
    return 'Negative'
  else:
    return 'Neutral'
```

```python
df_analysis['sentiment']=df_analysis['cleand_content'].apply(sentiment)
```

```python
df_analysis.head(5)
```

```python
df_analysis["Polarity"] = df_analysis["content"].apply(
    lambda x: TextBlob(str(x)).sentiment.polarity)
```

```python
df_analysis["Subjectivity"] = df_analysis["content"].apply(
    lambda x: TextBlob(str(x)).sentiment.subjectivity
)
```

**Note**                               
1:Polarity :Range: -1 to 1 (Used for Sentiment Classification.)                  
2:Subjectivity:Range: 0 to 1 (Used to understand the nature of the text "Fact or Opinion".)
